```python
# --- EJERCICIO 1: FORWARD PROPAGATION ---

import numpy as np

# Semilla para reproducibilidad de los números aleatorios
np.random.seed(42)

# Definición de la arquitectura del modelo
n_in, n_hidden, n_out = 2, 3, 1

# Inicialización Xavier / He para evitar el desvanecimiento del gradiente
W1 = np.random.randn(n_in, n_hidden) * np.sqrt(2.0 / n_in)
b1 = np.zeros((1, n_hidden))
W2 = np.random.randn(n_hidden, n_out) * np.sqrt(2.0 / n_hidden)
b2 = np.zeros((1, n_out))

# Función de activación Sigmoide
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

# Forward Propagation (Propagación hacia adelante)
X = np.array([[0.5, -0.2]])       # Entradas de prueba
z1 = np.dot(X, W1) + b1           # Suma ponderada en capa oculta
a1 = sigmoid(z1)                 # Activación de la capa oculta
z2 = np.dot(a1, W2) + b2          # Suma ponderada en capa de salida
y_hat = sigmoid(z2)               # Predicción final del modelo

# Resultados
print("Activación Oculta (a1):", a1)
print("Predicción Final (y_hat):", y_hat)
```
